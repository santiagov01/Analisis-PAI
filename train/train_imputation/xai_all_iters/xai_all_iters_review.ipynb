```python
import os
import json
from config import CFG, MODELS_CONFIG, setup_logger

from xai_utils import (
    run_shap_pipeline,
    run_permutation_pipeline,
    extract_and_save_robust_features
)
def load_json(file_path):
    with open(file_path, 'r') as f:
        return json.load(f)

CFG.xai_output_dir = f'{CFG.Root}/Resultados/all_xai_outputs_npk' 
log_file = os.path.join(CFG.xai_output_dir, "xai_all_iters.log")
logger = setup_logger(name="XAI_all_iters", log_file=log_file)

robust_features = {}
#iterar all_xai_outputs_npk
folder = '/export2/svargash/PAI-Quindio/Code/Analisis-PAI/Resultados/all_xai_outputs_npk'
for root, dirs, files in os.walk(folder):
    print(f'Procesando carpeta: {root}')
    print(f'Contiene subcarpetas: {dirs}')
    for dir in dirs:
        iteracion = dir.split('_')[-1]
        print(f'Procesando iteración: {iteracion}')
        #ruta al archivo shap
        shap_file = os.path.join(root, dir,
            'shap_results_summary.json')
        
        #cargar los datos de shap
        shap_results = load_json(shap_file)
        
        #cargar los datos de permutation
        perm_file = os.path.join(root, dir,
            'permutation_importance',
            'most_frequent_variables_80.json')
        perm_results = load_json(perm_file)
        
        CFG.xai_output_dir = f'{CFG.Root}/Resultados/all_xai_outputs_npk/iteracion_{iteracion}'
        robust_features[iteracion] = extract_and_save_robust_features(
        shap_results=shap_results,
        perm_results=perm_results,
        elements=CFG.elements_list,
        output_dir=CFG.xai_output_dir,
        logger=logger
        )
    dirs.clear()

print(robust_features)
common_robust_features = {}

iterations = list(robust_features.keys())

for element in CFG.elements_list:
    common_features = set(
        robust_features[iterations[0]].get(element, [])
    )

    for iteration in iterations[1:]:
        common_features.intersection_update(
            robust_features[iteration].get(element, [])
        )
        print(common_features)

    common_robust_features[element] = sorted(common_features)

path = os.path.join(
    CFG.Root,
    "Resultados",
    "all_xai_outputs_npk",
    "common_robust_features.json"
)

with open(path, "w") as f:
    json.dump(common_robust_features, f, indent=4, ensure_ascii=False)

print(common_robust_features)
```

```python
import os
import json
import math
from pathlib import Path
from collections import Counter

import pandas as pd


# Carpeta principal con iteracion_0, iteracion_1, etc.
root = Path(CFG.Root) / "Resultados" / "all_xai_outputs_cuartiles"

iteration_dirs = sorted(
    [
        directory
        for directory in root.iterdir()
        if directory.is_dir() and directory.name.startswith("iteracion_")
    ],
    key=lambda directory: int(directory.name.split("_")[-1])
)

threshold_percentage = 80
minimum_iterations = math.ceil(
    len(iteration_dirs) * threshold_percentage / 100
)

variable_counts = Counter()
processed_iterations = []

for iteration_dir in iteration_dirs:
    csv_path = iteration_dir / "cuartiles_best_vars_80.csv"

    if not csv_path.exists():
        print(f"No existe el archivo: {csv_path}")
        continue

    df = pd.read_csv(csv_path)

    if df.empty:
        continue

    # Usa la columna esperada; si no existe, usa la primera columna.
    column_name = (
        "Cuartiles_Best_Vars"
        if "Cuartiles_Best_Vars" in df.columns
        else df.columns[0]
    )

    # Set para que una variable cuente una sola vez por iteración.
    variables_in_iteration = {
        str(variable).strip()
        for variable in df[column_name].dropna()
        if str(variable).strip()
    }

    variable_counts.update(variables_in_iteration)
    processed_iterations.append(iteration_dir.name)


common_variables = sorted(
    variable
    for variable, count in variable_counts.items()
    if count >= minimum_iterations
)

common_robust_features = {
    "Cuartiles": common_variables
}

output_path = root / "common_cuartiles_best_vars_80.json"

with open(output_path, "w", encoding="utf-8") as file:
    json.dump(
        common_robust_features,
        file,
        indent=4,
        ensure_ascii=False
    )

print(f"Iteraciones procesadas: {len(processed_iterations)}")
print(f"Umbral requerido: {minimum_iterations} iteraciones")
print(f"Variables comunes: {common_variables}")
print(f"Resultado guardado en: {output_path}")
```
